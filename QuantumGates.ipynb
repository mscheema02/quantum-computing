### Code:

Imports

```python
import numpy as np

# Qiskit
from qiskit import QuantumCircuit, transpile, QuantumRegister, ClassicalRegister
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram, plot_state_city, plot_bloch_multivector
from qiskit.quantum_info import Statevector
import qiskit.quantum_info as qi
```

One qubit initialized to zero, passed through three gates: H then
Z then H; and then measured, along with state vector after each gate

```python
backend = AerSimulator()

q = QuantumRegister(1,'q')
c = ClassicalRegister(1,'c')

qc = QuantumCircuit(q,c)

# [1, 0] corresponds to |0>
qc.initialize([1,0], q)
qc.save_statevector(label='v0')

# Apply Hadamard+ Matrix
qc.h(q[0])
qc.save_statevector(label='v1')

# Apply Pauli Z Matrix
qc.z(q[0])
qc.save_statevector(label='v2')

# Apply Hadamard+ Matrix
qc.h(q[0])
qc.save_statevector(label='v3')

# Measure
qc.measure(q,c)

print(qc.draw(output='text'))

job = backend.run(qc, shots=1000)

result = job.result(timeout=10)

counts = result.get_counts(qc)
print(counts)

for i in range(0,4):
  print("At barrier "+ str(i) +":")
  print(np.asarray(result.data(0)['v'+str(i)]))
```
